# Matplotlib Object Oriented Method
Now that we've seen the basics, let's break it all down with a more formal introduction of Matplotlib's Object Oriented API. This means we will instantiate figure objects and then call methods or attributes from that object.

## Introduction to the Object Oriented Method

The main idea in using the more formal Object Oriented method is to create figure objects and then just call methods or attributes off of that object. This approach is nicer when dealing with a canvas that has multiple plots on it. 

To begin we create a figure instance. Then we can add axes to that figure:

```python
# Create Figure (empty canvas)
fig = plt.figure()

# Add set of axes to figure
axes = fig.add_axes([0.1, 0.1, 0.8, 0.8]) # left, bottom, width, height (range 0 to 1)

# Plot on that set of axes
axes.plot(x, y, 'b') #plots y and x in blue.
axes.set_xlabel('Set X Label') # sets the x axis label.
axes.set_ylabel('Set y Label') #sets the y axis label.
axes.set_title('Set Title') #sets the title.
```

Code is a little more complicated, but the advantage is that we now have full control of where the plot axes are placed, and we can easily add more than one axis to the figure:

```python
# Creates blank canvas
fig = plt.figure()

axes1 = fig.add_axes([0.1, 0.1, 0.8, 0.8]) # main axes
axes2 = fig.add_axes([0.2, 0.5, 0.4, 0.3]) # inset axes

# Larger Figure Axes 1
axes1.plot(x, y, 'b') # plot y and x in blue on the main axes
axes1.set_xlabel('X_label_axes1') # set x axis label for main plot
axes1.set_ylabel('Y_label_axes1') # set y axis label for main plot
axes1.set_title('Axes 1 Title') # set title for main plot

# Insert Figure Axes 2
axes2.plot(y, x, 'r') # plot x and y in red on the inset axes
axes2.set_xlabel('X_label_axes2') # set x axis label for inset plot
axes2.set_ylabel('Y_label_axes2') # set y axis label for inset plot
axes2.set_title('Axes 2 Title'); # set title for inset plot
```

## subplots()

The plt.subplots() object will act as a more automatic axis manager.

Basic use cases:

```python
# Use similar to plt.figure() except use tuple unpacking to grab fig and axes
fig, axes = plt.subplots()

# Now use the axes object to add stuff to plot
axes.plot(x, y, 'r') # plot y and x in red
axes.set_xlabel('x') # set x axis label
axes.set_ylabel('y') # set y axis label
axes.set_title('title'); # set plot title
```

Then you can specify the number of rows and columns when creating the subplots() object:

```python
# Empty canvas of 1 by 2 subplots
fig, axes = plt.subplots(nrows=1, ncols=2)
```

```python
# Axes is an array of axes to plot on
axes
```

We can iterate through this array:

```python
for ax in axes: # Loop
    ax.plot(x, y, 'b') # plot y and x in blue on each plot
    ax.set_xlabel('x') # set x axis label
    ax.set_ylabel('y') # set y axis label
    ax.set_title('title') # set title for each plot


# Display the figure object    
fig
```

A common issue with matplolib is overlapping subplots or figures. We ca use **fig.tight_layout()** or **plt.tight_layout()** method, which automatically adjusts the positions of the axes on the figure canvas so that there is no overlapping content:

```python
fig, axes = plt.subplots(nrows=1, ncols=2) # Create a figure with 1 row and 2 columns (two side-by-side subplots)

for ax in axes: # Loop
    ax.plot(x, y, 'g') # plot y and x in green
    ax.set_xlabel('x') # set x axis label
    ax.set_ylabel('y') # set y axis label
    ax.set_title('title') # set title for each plot

fig # display
plt.tight_layout() # adjusts spacing to prevent overlap
```

### Figure size, aspect ratio and DPI

Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the `figsize` and `dpi` keyword arguments. 
* `figsize` is a tuple of the width and height of the figure in inches
* `dpi` is the dots-per-inch (pixel per inch). 

For example:

```python
fig = plt.figure(figsize=(8,4), dpi=100) #Create a new figure with size 8x4 and resolution 100
```

The same arguments can also be passed to layout managers, such as the `subplots` function:

```python
fig, axes = plt.subplots(figsize=(12,3)) # create a figure and axes with size 12x3

axes.plot(x, y, 'r') # plot y and x in red
axes.set_xlabel('x') # set x axis label
axes.set_ylabel('y') # set y axis label
axes.set_title('title'); # set title
```

## Legends, labels and titles

Now that we have covered the basics of how to create a figure canvas and add axes instances to the canvas, let's look at how decorate a figure with titles, axis labels, and legends.

**Figure titles**

A title can be added to each axis instance in a figure. To set the title, use the `set_title` method in the axes instance:

```python
ax.set_title("title"); #set the title
```

**Axis labels**

Similarly, with the methods `set_xlabel` and `set_ylabel`, we can set the labels of the X and Y axes:

```python
ax.set_xlabel("x") # set the label for the x axis
ax.set_ylabel("y"); # set the label for the y axis
```

### Legends

You can use the **label="label text"** keyword argument when plots or other objects are added to the figure, and then using the **legend** method without arguments to add the legend to the figure: 

```python
fig = plt.figure() # create a new figure

ax = fig.add_axes([0,0,1,1]) # add axes for the figure


ax.plot(x, x**2, label="x**2") # plot x squared and assign it a label for the legend
ax.plot(x, x**3, label="x**3") # plot x cubed and assign it a label for the legend

ax.legend() # display the legend
```

Notice how are legend overlaps some of the actual plot!

The **legend** function takes an optional keyword argument **loc** that can be used to specify where in the figure the legend is to be drawn. The allowed values of **loc** are numerical codes for the various places the legend can be drawn. See the [documentation page](https://matplotlib.org/stable/) for details. Some of the most common **loc** values are:

```python
# Lots of options....

ax.legend(loc=1) # upper right corner
ax.legend(loc=2) # upper left corner
ax.legend(loc=3) # lower left corner
ax.legend(loc=4) # lower right corner

# .. many more options are available

# Most common to choose
ax.legend(loc=0) # let matplotlib decide the optimal location
fig
```

## Setting colors, linewidths, linetypes

Matplotlib gives you *a lot* of options for customizing colors, linewidths, and linetypes. 

There is the basic MATLAB like syntax (which I would suggest you avoid using for more clairty sake:

### Colors with MatLab like syntax

With matplotlib, we can define the colors of lines and other graphical elements in a number of ways. First of all, we can use the MATLAB-like syntax where `'b'` means blue, `'g'` means green, etc. The MATLAB API for selecting line styles are also supported: where, for example, 'b.-' means a blue line with dots:

```python
# MATLAB style line color and style 
fig, ax = plt.subplots()
ax.plot(x, x**2, 'b.-') # blue line with dots
ax.plot(x, x**3, 'g--') # green dashed line
```

### Colors with the color= parameter

We can also define colors by their names or RGB hex codes and optionally provide an alpha value using the `color` and `alpha` keyword arguments. Alpha indicates opacity.

```python
fig, ax = plt.subplots() #create a new figure

ax.plot(x, x+1, color="blue", alpha=0.5) # half-transparant
ax.plot(x, x+2, color="#8B008B")        # RGB hex code
ax.plot(x, x+3, color="#FF8C00")        # RGB hex code 
```

### Line and marker styles

To change the line width, we can use the `linewidth` or `lw` keyword argument. The line style can be selected using the `linestyle` or `ls` keyword arguments:

```python
fig, ax = plt.subplots(figsize=(12,6)) #create a figure

ax.plot(x, x+1, color="red", linewidth=0.25) # Plot with thin red line
ax.plot(x, x+2, color="red", linewidth=0.50) # Slightly thicker red line
ax.plot(x, x+3, color="red", linewidth=1.00) # Medium red line
ax.plot(x, x+4, color="red", linewidth=2.00) # Thick red line

# possible linestype options ‘-‘, ‘–’, ‘-.’, ‘:’, ‘steps’
ax.plot(x, x+5, color="green", lw=3, linestyle='-') # Green solid line 
ax.plot(x, x+6, color="green", lw=3, ls='-.') # Green dash dot line
ax.plot(x, x+7, color="green", lw=3, ls=':') # Green dotted line


# custom dash
line, = ax.plot(x, x+8, color="black", lw=1.50) # black line saved as variable to modify
# Custom dash pattern
line.set_dashes([5, 10, 15, 10]) # format: line length, space length, ...

# possible marker symbols: marker = '+', 'o', '*', 's', ',', '.', '1', '2', '3', '4', ...
ax.plot(x, x+ 9, color="blue", lw=3, ls='-', marker='+') # blue solid line with plus markers
ax.plot(x, x+10, color="blue", lw=3, ls='--', marker='o') # blue dashed line with circle markers
ax.plot(x, x+11, color="blue", lw=3, ls='-', marker='s') # blue solid line with square markers
ax.plot(x, x+12, color="blue", lw=3, ls='--', marker='1') # blue dashed line with triangle-down marker


# marker size and color
ax.plot(x, x+13, color="purple", lw=1, ls='-', marker='o', markersize=2) # small circle markers
ax.plot(x, x+14, color="purple", lw=1, ls='-', marker='o', markersize=4) # medium circle markers
ax.plot(x, x+15, color="purple", lw=1, ls='-', marker='o', markersize=8, markerfacecolor="red") # large circles with red fill
ax.plot(x, x+16, color="purple", lw=1, ls='-', marker='s', markersize=8, 
        markerfacecolor="yellow", markeredgewidth=3, markeredgecolor="green"); # square markers with yellow fill and green border
```

### Control over axis appearance

In this section we will look at controlling axis sizing properties in a matplotlib figure.

## Plot range

We can configure the ranges of the axes using the `set_ylim` and `set_xlim` methods in the axis object, or `axis('tight')` for automatically getting "tightly fitted" axes ranges:

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4)) # create a figure with 1 row and 3 plots


axes[0].plot(x, x**2, x, x**3) # plot x squared and x cubed on the first plot

axes[0].set_title("default axes ranges") # set title

axes[1].plot(x, x**2, x, x**3) # plot x squared and x cubed on the second plot
axes[1].axis('tight') # adjusts axis to fit the data

axes[1].set_title("tight axes") # set title for second plot

axes[2].plot(x, x**2, x, x**3) # plot x squared and x cubed on the third plot
axes[2].set_ylim([0, 60]) # set y axis range
axes[2].set_xlim([2, 5]) # set custom x axis range
axes[2].set_title("custom axes range"); # set title for third subplot
```
